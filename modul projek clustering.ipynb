```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

STANDARISASI

```python
df= pd.read_csv(r"path", sep="sesuaiakn separator csv nya") #  <-- masukkan path file csv yang sudah melalui preprocessing dan feature engineering

kecamatan= df["Kecamatan"] #  <-- sesuaikan dengan nama kolom kecamatan

kolom_numerik= df.select_dtypes(include="number").columns

#proses standarisasi
scaler= MinMaxScaler()
df_scaled= pd.DataFrame(scaler.fit_transform(df[kolom_numerik]), columns=kolom_numerik)

df_scaled.insert(0, "Kecamatan", kecamatan)
df_scaled
```

ELBOW METHOD (MENENTUKAN JUMLAH CLUSTER)

```python
X = df_scaled.select_dtypes(include='number').values

inertia = []
K = range(1, 11)

for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    inertia.append(km.inertia_)

plt.figure()
plt.plot(K, inertia, marker='o')
plt.xlabel("Jumlah Cluster (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid()
plt.show()
```

PROSES CLUSTERING

```python
X = df_scaled.select_dtypes(include='number')

kmeans = KMeans(n_clusters=3, random_state=42)
df["Cluster"] = kmeans.fit_predict(X)

df
```

VISUALISASI HASIL CLUSTERING

```python
X = df_scaled.select_dtypes(include='number')

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

df["PCA1"] = X_pca[:, 0]
df["PCA2"] = X_pca[:, 1]

plt.figure(figsize=(10, 7))

for cluster in sorted(df["Cluster"].unique()):
    subset = df[df["Cluster"] == cluster]
    plt.scatter(
        subset["PCA1"],
        subset["PCA2"],
        label=f"Cluster {cluster}"
    )

for i, row in df.iterrows():
    plt.text(
        row["PCA1"],
        row["PCA2"],
        row["Kecamatan"],
        fontsize=8,
        alpha=0.8
    )

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Visualisasi Cluster Kecamatan dengan Label")
plt.legend()
plt.grid()
plt.show()
```
